```python
import pandas as pd

views = pd.read_csv('../data/feed-views.log', delimiter='\t',
 skiprows=[2,3], skipfooter=2, names = ['datetime', 'user'], engine ='python')

views['datetime'] = pd.to_datetime(views['datetime'], format='%Y-%m-%d %H:%M:%S.%f')
print(views['datetime'].dtype)

views['year'] = views['datetime'].dt.year
views['month'] = views['datetime'].dt.month
views['day'] = views['datetime'].dt.day
views['hour'] = views['datetime'].dt.hour
views['minute'] = views['datetime'].dt.minute
views['second'] = views['datetime'].dt.second

bins = [0, 4, 7, 11, 17, 20, 24]
labels = ['night', 'early morning', 'morning', 'afternoon', 'early evening', 'evening']

views['daytime'] = pd.cut(views['hour'], bins=bins, labels=labels, right=False, include_lowest=True)
print(f'\ndaytimes: {views["daytime"]}')

views = views.set_index('user')

total_count = views.count()
daytime_counts = views['daytime'].value_counts()
print(f"\ndaytime count: {daytime_counts}")

views_sorted = views.sort_values(['hour', 'minute', 'second'])
# print(f"\nview sorted: {views_sorted}")

min_hour_morning = views[views['daytime']=='morning']['hour'].min()
max_hour_night = views[views['daytime']=='night']['hour'].max()
print(f"\nmin and max hours: {min_hour_morning} {max_hour_night}")

night_users = views[views['daytime']=='night']
morning_users = views[views['daytime']=='morning']

example_night_user = night_users[night_users['hour']==max_hour_night].index[0]
example_morning_user = morning_users[morning_users['hour']==min_hour_morning].index[0]

daytime_mode = views['daytime'].mode()[0]
hour_mode = views['hour'].mode()[0]
print(f"moda: {daytime_mode}")

earliest_hour = views.nsmallest(3, ['hour', 'minute', 'second'])[['hour', 'minute', 'second']]
latest_hour = views.nlargest(3, ['hour', 'minute', 'second'])[['hour', 'minute', 'second']]
print(f"\nearliest hour: {earliest_hour}")
print(f"\nlatest hour: {latest_hour}")

earliest_with_users = views.nsmallest(3, ['hour', 'minute', 'second'])
latest_with_users = views.nlargest(3, ['hour', 'minute', 'second'])

#стата по интервалам
stats = views['hour'].describe()
print(f"{stats}")

q1 = stats['25%']
q3 = stats['75%']
iqr = q3 - q1
print(f"\niqr for hour: {iqr}")
print(f"most popular visitin interval: {q1} to {q3}")
```
