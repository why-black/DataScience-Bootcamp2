```python
import pandas as pd

df = pd.read_csv("../data/auto.csv", delimiter=',', index_col='ID')

observation_count = df.count()
print(f"observation count: {observation_count}")

df = df.drop_duplicates(subset=['CarNumber', 'Make_n_model', 'Fines'], keep='last')
observation_cleared_count = df.count()
print(f"after clear count: {observation_cleared_count}")

missing_values = df.isna().sum()
print(f"missing values: {missing_values}")

df = df.dropna(axis=1, thresh=len(df)-500)

df['Refund'] = df['Refund'].fillna(method='ffill')
df['Fines'] = df['Fines'].fillna(df['Fines'].mean())

df['Make'] = df['Make_n_model'].apply(lambda x: x.split()[0] if pd.notna(x) else None)
df['Model'] = df['Make_n_model'].apply(lambda x: ' '.join(x.split()[1:]) if pd.notna(x) else None)

df = df.drop('Make_n_model', axis=1)
df.to_json('../data/auto.json', orient='records', indent=2)
```
