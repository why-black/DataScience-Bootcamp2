```python
import pandas as pd

df = pd.read_json("../data/auto.json")
df = df.set_index('CarNumber')

fines_more_2100 = df[df['Fines'] > 2100]
print(f"fines more than 2100: {fines_more_2100}")

more_2100_refund_2 = fines_more_2100[fines_more_2100['Refund'] == 2]
print(f"\nfines more 2100 and refund is 2: {more_2100_refund_2}")

model_request = ['Focus', 'Corolla']
focus_corolla = df[df['Model'].isin(model_request)]
print(f"\nonly focus and corolla: {focus_corolla}")

numbers_list = ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
cars_filter = df[df.index.isin(numbers_list)]
print(f"\nfiltered car by numbers: {cars_filter}")

median_by_make = df.groupby('Make')['Fines'].median()
print(f"\nmedian grouped by make: {median_by_make.sort_values(ascending=False)}")

median_by_make_model = df.groupby(['Make', 'Model'])['Fines'].median()
print(f"\nmedian grouped by make and model: {median_by_make_model.sort_values(ascending=False)}")

count_by_make_model = df.groupby(['Make', 'Model'])['Fines'].count()
print(f"\nnums of fines by make and model: {count_by_make_model.sort_values(ascending=False)}")

min_max = df.groupby(['Make', 'Model'])['Fines'].agg(['min', 'max'])
print(f"\nmin and max: {min_max}")

std_dev = df.groupby(['Make', 'Model'])['Fines'].std()
print(f"\nstandart deviation: {std_dev}")

fines_count_by_car = df.groupby(df.index)['Fines'].count()
print("number of fines per car (descending):")
print(fines_count_by_car.sort_values(ascending=False).head(10))

top_car_by_count = fines_count_by_car.idxmax()
print(f"\ncar with most fines: {top_car_by_count}")
top_car_data = df.loc[top_car_by_count]
print(f"data for {top_car_by_count}:")
print(top_car_data)

fines_sum_by_car = df.groupby(df.index)['Fines'].sum()
print("\ntotal fines per car (descending):")
print(fines_sum_by_car.sort_values(ascending=False).head(10))

top_car_by_sum = fines_sum_by_car.idxmax()
top_car_sum_data = df.loc[top_car_by_sum]
print(f"data for {top_car_by_sum}:")
print(top_car_sum_data)

models_per_car = df.groupby(df.index)['Model'].nunique()
cars_with_multiple_models = models_per_car[models_per_car > 1]
print("Cars with multiple models:")
print(cars_with_multiple_models)

if len(cars_with_multiple_models) > 0:
    for car in cars_with_multiple_models.index:
        car_data = df.loc[car]
        print(car_data[['Model', 'Make']].drop_duplicates())
else:
    print("No cars are connected to different models.")
```
